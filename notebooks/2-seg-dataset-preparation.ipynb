```python
from glob import glob
import cv2
from tqdm.auto import tqdm
```

```python
import cell_contraction_monitor.processing.wells_selection as ccm_ws
import cell_contraction_monitor.io.utils as io_utils
```

# Build dataset

```python
import os
TARGET_SHAPE = 348
image_mask = "*.JPG"
src_dataset_root = "Z:/Статьи/Cell Contraction/raw-data"
src_dataset_mask = os.path.join(src_dataset_root, "*", image_mask)
target_path = "Z:/Статьи/Cell Contraction/segmentation_dataset"
```

Print source data structure

```python
all_images = glob(src_dataset_mask)
experiments = set()
for img_path in all_images:
    dir_name = os.path.dirname(img_path)
    experiments.add(os.path.split(dir_name)[1])

print(f"Found {len(all_images)} images in {len(experiments)} experiments")
```

We will take only 5 images from each experiment to keep dataset reasonable size

```python
for experiment in tqdm(experiments):
    exp_images = sorted(
        glob(os.path.join(src_dataset_root, experiment, image_mask)))
    print(f"Process experiment {experiment} with {len(exp_images)} images")
    images_to_segment = []
    images_to_segment.append(exp_images[0])
    images_to_segment.append(exp_images[len(exp_images)//4])
    images_to_segment.append(exp_images[len(exp_images)//2])
    images_to_segment.append(exp_images[len(3*exp_images)//4])
    images_to_segment.append(exp_images[-1])

    for img_path in tqdm(images_to_segment, leave=False):
        image_np = io_utils.read_image(img_path)
        if len(image_np.shape) == 3:
            image_np = cv2.cvtColor(image_np, cv2.COLOR_BGR2GRAY)

        centers, average_radius = ccm_ws.find_wells(image_np)
        well_images, well_masks = ccm_ws.get_well_images(
            image_np, centers, average_radius)
        for i, well_image in tqdm(enumerate(well_images), leave=False):
            target_image_path = os.path.join(target_path, experiment, os.path.split(
                img_path)[1].replace('.JPG', f'_{i}.JPG'))
            os.makedirs(os.path.dirname(target_image_path), exist_ok=True)
            border_size = TARGET_SHAPE - well_image.shape[0]
            if border_size % 2 == 1:
                top = border_size // 2
                bottom = top + 1
            else:
                top = bottom = border_size // 2

            well_image = cv2.copyMakeBorder(
                well_image, top, bottom, top, bottom, cv2.BORDER_CONSTANT, None, 0)
            io_utils.write_image(target_image_path, well_image)
```
